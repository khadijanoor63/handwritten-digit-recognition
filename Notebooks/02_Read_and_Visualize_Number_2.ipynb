```python
im = pd.read_csv(path + "2.csv", header=None, skiprows=None)
t1 = im.iloc[0,1:]
t1 = np.array(t1)
t2 = t1.reshape(28,28)
fig = plt.figure(figsize=(3,3))
plt.imshow(t2, cmap="gray")
plt.show()
```

```python
pd.options.display.max_columns = None
pd.options.display.max_rows = None
pd.DataFrame(t2)
```

```python
X = copy.deepcopy(t2)
Y = copy.deepcopy(t2)
Z = copy.deepcopy(t2)

for i in range (len(t2)):
  for j in range (len(t2[i])):
    if t2[i][j] >= 0:
      X[i][j]= i
      Y[i][j]= j
```

```python
%matplotlib notebook

fig = plt.figure()
ax = plt.axes(projection = '3d')
ax.scatter(X,Y,Z, c=Z, cmap = 'viridis', linewidth=0.5)

#Not able to see 3d view
```

```python
#Single sample Surface plot
import plotly.graph_objects as go
fig = go.Figure(data=[go.Surface(z=Z, x=X, y=Y, colorscale = 'Plasma')])
fig.update_layout(
    title='Surface Plot for the data representation of 2',
    scene = dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z'
    )
)

fig.show()
```
