```python
import cv2
from matplotlib import pyplot as plt
import pandas as pd
import numpy as np
import csv
```

```python
from google.colab import drive
drive.mount('/content/drive')
#upload from drive
```

```python
image_path = "/content/drive/MyDrive/Data/"
```

```python
# Load the image
letter_B = cv2.imread(image_path + "B.jpg", cv2.IMREAD_GRAYSCALE)

print("Displaying the image that is read:")

fig = plt.figure(figsize=(3,3))
plt.imshow(letter_B, cmap="gray")
plt.show()
```

```python
#convert the image to a grid of numbers (data)
pd.options.display.max_columns = None
pd.options.display.max_rows = None
pd.DataFrame(letter_B)
```

```python
#Flatten data
flattened_B = sum((letter_B).tolist(),[])
flattened_B
```

```python
#Resize data to original image
t1 = np.array(flattened_B)
t2 = t1.reshape(33,24)
fig = plt.figure(figsize=(3,3))
plt.imshow(t2, cmap="gray")
plt.show()
```

```python
#creating a csv file of flattened data and adding it to the folder where the image is
filepath = image_path
with open(filepath + 'image_to_numbers.csv', 'w', newline='') as new_file:
  write = csv.writer(new_file)
  write.writerow(map(lambda x: [x][0], flattened_B))
```
